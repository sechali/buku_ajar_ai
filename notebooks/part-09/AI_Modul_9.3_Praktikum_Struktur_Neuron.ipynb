```python
import numpy as np
import matplotlib.pyplot as plt

plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
print(f'NumPy Version: {np.__version__}')
```

## 1. Implementasi Kelas Single Perceptron Rosenblatt
Membuat kelas Perceptron dari awal dengan fungsi tangga Heaviside dan pembaruan bobot berbasis galat biner.

```python
class RosenblattPerceptron:
    def __init__(self, n_features, learning_rate=0.1, max_epochs=50, seed=42):
        np.random.seed(seed)
        self.w = np.zeros(n_features)
        self.b = 0.0
        self.lr = learning_rate
        self.max_epochs = max_epochs
        self.history_errors = []
        
    def step_function(self, z):
        return np.where(z >= 0.0, 1, 0)
        
    def predict(self, X):
        z = np.dot(X, self.w) + self.b
        return self.step_function(z)
        
    def fit(self, X, y):
        m = X.shape[0]
        for epoch in range(self.max_epochs):
            total_error = 0
            for i in range(m):
                y_hat = self.predict(X[i])
                error = y[i] - y_hat
                if error != 0:
                    self.w += self.lr * error * X[i]
                    self.b += self.lr * error
                    total_error += abs(error)
            self.history_errors.append(total_error)
            if total_error == 0:
                break
        return self

print('Kelas RosenblattPerceptron berhasil didefinisikan!')
```

## 2. Pengujian pada Gerbang Linier (AND) vs Gerbang Non-Linier (XOR)
Menguji keterbatasan matematis perceptron tunggal terhadap gerbang logika biner.

```python
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y_and = np.array([0, 0, 0, 1])
y_xor = np.array([0, 1, 1, 0])

# Training pada Gerbang AND
p_and = RosenblattPerceptron(n_features=2, learning_rate=0.2, max_epochs=20)
p_and.fit(X, y_and)
print('=== HASIL GERBANG AND (Linearly Separable) ===')
print('Bobot Akhir w   :', p_and.w)
print('Bias Akhir b    :', p_and.b)
print('Prediksi Output :', p_and.predict(X))
print('Konvergen Epoch :', len(p_and.history_errors))

# Training pada Gerbang XOR
p_xor = RosenblattPerceptron(n_features=2, learning_rate=0.2, max_epochs=30)
p_xor.fit(X, y_xor)
print('\n=== HASIL GERBANG XOR (Non-Linearly Separable) ===')
print('Bobot Akhir w   :', p_xor.w)
print('Bias Akhir b    :', p_xor.b)
print('Prediksi Output :', p_xor.predict(X))
print('Total Error Akhir:', p_xor.history_errors[-1])
```

## 3. Visualisasi Dinamika Galat Perceptron pada Gerbang XOR
Menampilkan kegagalan konvergensi berupa osilasi galat tak berujung pada data non-linier XOR.

```python
fig, ax = plt.subplots(figsize=(8, 4.5))
ax.plot(p_and.history_errors, marker='o', label='Gerbang AND (Konvergen ke 0)', color='forestgreen', linewidth=2)
ax.plot(p_xor.history_errors, marker='s', label='Gerbang XOR (Osilasi / Gagal Konvergen)', color='crimson', linewidth=2)
ax.set_title('Dinamika Konvergensi Perceptron: AND vs XOR', fontsize=12, fontweight='bold')
ax.set_xlabel('Epoch Iterasi', fontsize=10)
ax.set_ylabel('Jumlah Kesalahan Klasifikasi', fontsize=10)
ax.legend()
plt.tight_layout()
plt.savefig('praktikum_and_vs_xor_9_3.png', dpi=150)
plt.show()
```
